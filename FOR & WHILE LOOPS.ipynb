# WHILE LOOP:

```python
i = 1

while i <= 15:
    print(i)
    i += 1
```

```python
num = int(input("Enter a number: "))

i = 1
while i <= 10:
    print(num, "x", i, "=", num * i)
    i += 1
```

```python
total = 0

num = int(input("Enter a number (0 to stop): "))

while num != 0:
    total += num
    num = int(input("Enter a number (0 to stop): "))

print("Sum =", total)
```

```python
i = 1

while i <= 20:
    print(i)
    i += 2
```

```python
num = float(input("Enter a positive number: "))

count = 0

while num >= 1:
    num = num / 2
    count += 1

print("Steps =", count)
```

```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number =", reverse)
```

```python
text = ""

while text.lower() != "stop":
    text = input("Enter something (type 'stop' to end): ")

print("Program Ended")
```

```python
a = 0
b = 1
count = 1

while count <= 10:
    print(a)
    c = a + b
    a = b
    b = c
    count += 1
```

```python
num = int(input("Enter a number: "))

i = 2
is_prime = True

while i < num:
    if num % i == 0:
        is_prime = False
        break
    i += 1

if num <= 1:
    print("Not Prime")
elif is_prime:
    print("Prime Number")
else:
    print("Not Prime")
```

```python
while True:
    print("\nMenu")
    print("1. Add")
    print("2. Subtract")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Addition =", a + b)

    elif choice == 2:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Subtraction =", a - b)

    elif choice == 3:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```

# HARD QUESTIONS FOR & WHILE:

```python
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
n = int(input("Enter a number: "))

num = 2

while num <= n:
    i = 2
    is_prime = True

    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print(num)

    num += 1
```

```python
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial *= i

print("Factorial =", factorial)
```

```python
start = int(input("Enter start number: "))
end = int(input("Enter end number: "))

num = start

while num <= end:
    temp = num
    total = 0
    digits = len(str(num))

    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10

    if total == num:
        print(num)

    num += 1
```

```python
import random

secret = random.randint(1, 100)

chance = 1

while chance <= 7:
    guess = int(input("Enter your guess: "))

    if guess == secret:
        print("Congratulations! You guessed the number.")
        break

    elif guess < secret:
        print("Too Low")

    else:
        print("Too High")

    chance += 1

if chance > 7:
    print("Game Over!")
    print("The correct number was:", secret)
```
