```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/user/Downloads/matplotlib_sales_clean_1000 - matplotlib_sales_clean_1000.csv")

city_profit = (
    df.groupby("City")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

bars = plt.bar(city_profit.index, city_profit.values)

highest_city = city_profit.idxmax()
highest_profit = city_profit.max()

highest_index = list(city_profit.index).index(highest_city)

bars[highest_index].set_alpha(0.5)

plt.annotate(
    f"Highest: {highest_city}\n{highest_profit:.2f}",
    xy=(highest_index, highest_profit),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

plt.title("Total Profit by City")
plt.xlabel("City")
plt.ylabel("Total Profit")
plt.grid(axis="y")

plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/user/Downloads/matplotlib_sales_clean_1000 - matplotlib_sales_clean_1000.csv")

columns = [
    "Customer_Age",
    "Units_Sold",
    "Unit_Price",
    "Discount_Pct",
    "Revenue",
    "Cost",
    "Profit"
]

corr = df[columns].corr()

plt.figure(figsize=(10, 8))

plt.imshow(corr)

plt.xticks(
    range(len(columns)),
    columns,
    rotation=45
)

plt.yticks(
    range(len(columns)),
    columns
)

plt.colorbar(label="Correlation")

for i in range(len(columns)):
    for j in range(len(columns)):

        plt.text(
            j,
            i,
            f"{corr.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/user/Downloads/matplotlib_sales_clean_1000 - matplotlib_sales_clean_1000.csv")

df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Month"] = df["Order_Date"].dt.strftime("%Y-%m")

monthly = df.groupby("Month")[["Revenue", "Profit"]].sum()
category = df.groupby("Product_Category")["Revenue"].sum()
city = df.groupby("City")["Profit"].sum()

highest_month = monthly["Revenue"].idxmax()
highest_revenue = monthly["Revenue"].max()

fig, axes = plt.subplots(
    3,
    2,
    figsize=(16, 14)
)

# Monthly Revenue
axes[0, 0].plot(
    monthly.index,
    monthly["Revenue"],
    marker="o",
    label="Revenue"
)

axes[0, 0].set_title("Monthly Revenue")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue")
axes[0, 0].legend()
axes[0, 0].grid(True)
axes[0, 0].tick_params(axis="x", rotation=45)

axes[0, 0].annotate(
    f"Highest: {highest_month}",
    xy=(highest_month, highest_revenue),
    xytext=(0, 25),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

# Monthly Profit
axes[0, 1].plot(
    monthly.index,
    monthly["Profit"],
    marker="o",
    label="Profit"
)

axes[0, 1].set_title("Monthly Profit")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Profit")
axes[0, 1].legend()
axes[0, 1].grid(True)
axes[0, 1].tick_params(axis="x", rotation=45)

# Revenue by Category
axes[1, 0].bar(
    category.index,
    category.values
)

axes[1, 0].set_title("Revenue by Category")
axes[1, 0].set_xlabel("Product Category")
axes[1, 0].set_ylabel("Revenue")
axes[1, 0].grid(axis="y")
axes[1, 0].tick_params(axis="x", rotation=45)

# Profit by City
axes[1, 1].bar(
    city.index,
    city.values
)

axes[1, 1].set_title("Profit by City")
axes[1, 1].set_xlabel("City")
axes[1, 1].set_ylabel("Profit")
axes[1, 1].grid(axis="y")
axes[1, 1].tick_params(axis="x", rotation=45)

# Revenue Distribution
axes[2, 0].hist(
    df["Revenue"],
    bins=20,
    edgecolor="black"
)

axes[2, 0].set_title("Revenue Distribution")
axes[2, 0].set_xlabel("Revenue")
axes[2, 0].set_ylabel("Frequency")
axes[2, 0].grid(axis="y")

# Discount vs Profit
axes[2, 1].scatter(
    df["Discount_Pct"],
    df["Profit"],
    alpha=0.6
)

axes[2, 1].set_title("Discount vs Profit")
axes[2, 1].set_xlabel("Discount %")
axes[2, 1].set_ylabel("Profit")
axes[2, 1].grid(True)

fig.suptitle(
    "Complete Sales Performance Dashboard",
    fontsize=18
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/user/Downloads/matplotlib_sales_clean_1000 - matplotlib_sales_clean_1000.csv")

# Convert date
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Create month
df["Month"] = df["Order_Date"].dt.strftime("%Y-%m")

# -------------------------
# Calculations
# -------------------------

monthly = df.groupby("Month")[["Revenue", "Profit"]].sum()

category_profit = (
    df.groupby("Product_Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

city_revenue = (
    df.groupby("City")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

segment_revenue = (
    df.groupby("Customer_Segment")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

top5_profit = (
    df.nlargest(5, "Profit")
    .sort_values("Profit")
)

highest_month = monthly["Revenue"].idxmax()
highest_month_revenue = monthly["Revenue"].max()

highest_category = category_profit.idxmax()
highest_category_profit = category_profit.max()

best_city = city_revenue.idxmax()
best_city_revenue = city_revenue.max()

best_segment = segment_revenue.idxmax()
best_segment_revenue = segment_revenue.max()

discount_profit_corr = df[
    ["Discount_Pct", "Profit"]
].corr().iloc[0, 1]

# -------------------------
# Dashboard
# -------------------------

fig, axes = plt.subplots(
    3,
    2,
    figsize=(17, 15)
)

# 1. Monthly Revenue

axes[0, 0].plot(
    monthly.index,
    monthly["Revenue"],
    marker="o"
)

axes[0, 0].set_title("Monthly Revenue")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue")
axes[0, 0].grid(True)
axes[0, 0].tick_params(axis="x", rotation=45)

axes[0, 0].annotate(
    f"Highest: {highest_month}",
    xy=(highest_month, highest_month_revenue),
    xytext=(0, 25),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

# 2. Profit by Product Category

axes[0, 1].bar(
    category_profit.index,
    category_profit.values
)

axes[0, 1].set_title("Profit by Product Category")
axes[0, 1].set_xlabel("Product Category")
axes[0, 1].set_ylabel("Profit")
axes[0, 1].grid(axis="y")
axes[0, 1].tick_params(axis="x", rotation=45)

axes[0, 1].annotate(
    f"Highest: {highest_category}",
    xy=(highest_category, highest_category_profit),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

# 3. Revenue by City

axes[1, 0].bar(
    city_revenue.index,
    city_revenue.values
)

axes[1, 0].set_title("Revenue by City")
axes[1, 0].set_xlabel("City")
axes[1, 0].set_ylabel("Revenue")
axes[1, 0].grid(axis="y")
axes[1, 0].tick_params(axis="x", rotation=45)

axes[1, 0].annotate(
    f"Best City: {best_city}",
    xy=(best_city, best_city_revenue),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

# 4. Revenue by Customer Segment

axes[1, 1].bar(
    segment_revenue.index,
    segment_revenue.values
)

axes[1, 1].set_title("Revenue by Customer Segment")
axes[1, 1].set_xlabel("Customer Segment")
axes[1, 1].set_ylabel("Revenue")
axes[1, 1].grid(axis="y")

axes[1, 1].annotate(
    f"Top Segment: {best_segment}",
    xy=(best_segment, best_segment_revenue),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

# 5. Discount vs Profit

axes[2, 0].scatter(
    df["Discount_Pct"],
    df["Profit"],
    alpha=0.6
)

axes[2, 0].set_title("Discount vs Profit")
axes[2, 0].set_xlabel("Discount Percentage")
axes[2, 0].set_ylabel("Profit")
axes[2, 0].grid(True)

axes[2, 0].annotate(
    f"Correlation = {discount_profit_corr:.2f}",
    xy=(0.05, 0.90),
    xycoords="axes fraction"
)

# 6. Top 5 Profit Orders

axes[2, 1].barh(
    top5_profit["Order_ID"].astype(str),
    top5_profit["Profit"]
)

axes[2, 1].set_title("Top 5 Orders by Profit")
axes[2, 1].set_xlabel("Profit")
axes[2, 1].set_ylabel("Order ID")
axes[2, 1].grid(axis="x")

fig.suptitle(
    "Business Analytics Sales Dashboard",
    fontsize=20
)

plt.tight_layout()
plt.show()

# -------------------------
# Business Findings
# -------------------------

print("========== BUSINESS FINDINGS ==========")

print(
    f"1. Highest Revenue Month: "
    f"{highest_month} ({highest_month_revenue:.2f})"
)

print(
    f"2. Highest Profit Category: "
    f"{highest_category} ({highest_category_profit:.2f})"
)

print(
    f"3. Best Performing City: "
    f"{best_city} ({best_city_revenue:.2f} revenue)"
)

print(
    f"4. Highest Revenue Customer Segment: "
    f"{best_segment} ({best_segment_revenue:.2f})"
)

print(
    f"5. Discount vs Profit Correlation: "
    f"{discount_profit_corr:.2f}"
)

print(
    "6. Revenue Distribution: "
    "Use the histogram/distribution to determine whether "
    "revenue is symmetric or skewed."
)

print("\n7. Top 5 Orders by Profit:")

print(
    top5_profit[
        ["Order_ID", "Revenue", "Profit"]
    ].to_string(index=False)
)
```
