```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "/Users/user/Downloads/matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv"
)

# Convert Order Date to datetime
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

# Convert Revenue to numeric
df["Revenue"] = pd.to_numeric(
    df["Revenue"],
    errors="coerce"
)

# Calculate total revenue for each month
monthly_revenue = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Revenue"].sum()

months = monthly_revenue.index.astype(str)

# Line chart
plt.figure(figsize=(10, 5))

plt.plot(
    months,
    monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.grid(True)

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "/Users/user/Downloads/matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv"
)

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Profit"] = pd.to_numeric(
    df["Profit"],
    errors="coerce"
)

# Monthly profit
monthly_profit = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Profit"].sum()

months = monthly_profit.index.astype(str)

plt.figure(figsize=(10, 5))

plt.plot(
    months,
    monthly_profit.values,
    linestyle="--",
    marker="s",
    label="Monthly Profit"
)

# Find highest-profit month
highest_month = monthly_profit.idxmax()
highest_profit = monthly_profit.max()

# Annotate highest profit
plt.annotate(
    f"Highest = {highest_profit:.2f}",
    xy=(str(highest_month), highest_profit),
    xytext=(10, 20),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.legend()
plt.grid(True)

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv(
    "/Users/user/Downloads/matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv"
)

df["Revenue"] = pd.to_numeric(
    df["Revenue"],
    errors="coerce"
)

df["Profit"] = pd.to_numeric(
    df["Profit"],
    errors="coerce"
)

category_data = df.groupby(
    "Product Category"
)[["Revenue", "Profit"]].sum()

categories = category_data.index

revenue = category_data["Revenue"]
profit = category_data["Profit"]

x = np.arange(len(categories))

width = 0.35

plt.figure(figsize=(10, 6))

plt.bar(
    x - width / 2,
    revenue,
    width,
    label="Revenue"
)

plt.bar(
    x + width / 2,
    profit,
    width,
    label="Profit"
)

plt.title("Revenue vs Profit by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Amount")

plt.xticks(
    x,
    categories,
    rotation=45
)

plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "/Users/user/Downloads/matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv"
)

# Convert required columns
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Revenue"] = pd.to_numeric(
    df["Revenue"],
    errors="coerce"
)

df["Profit"] = pd.to_numeric(
    df["Profit"],
    errors="coerce"
)

df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

# Monthly revenue
monthly_revenue = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Revenue"].sum()

# Category revenue
category_revenue = df.groupby(
    "Product Category"
)["Revenue"].sum()


# Create 2 x 2 dashboard
fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)


# -------------------------
# Chart 1: Monthly Revenue
# -------------------------

axes[0, 0].plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o"
)

axes[0, 0].set_title("Monthly Revenue")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue")
axes[0, 0].grid(True)

axes[0, 0].tick_params(
    axis="x",
    rotation=45
)


# -------------------------
# Chart 2: Revenue by Category
# -------------------------

axes[0, 1].bar(
    category_revenue.index,
    category_revenue.values
)

axes[0, 1].set_title("Revenue by Product Category")
axes[0, 1].set_xlabel("Product Category")
axes[0, 1].set_ylabel("Revenue")

axes[0, 1].tick_params(
    axis="x",
    rotation=45
)


# -------------------------
# Chart 3: Customer Age
# -------------------------

axes[1, 0].hist(
    df["Age"].dropna(),
    bins=10
)

axes[1, 0].set_title("Customer Age Distribution")
axes[1, 0].set_xlabel("Age")
axes[1, 0].set_ylabel("Frequency")
axes[1, 0].grid(True)


# -------------------------
# Chart 4: Revenue vs Profit
# -------------------------

axes[1, 1].scatter(
    df["Revenue"],
    df["Profit"],
    alpha=0.6
)

axes[1, 1].set_title("Revenue vs Profit")
axes[1, 1].set_xlabel("Revenue")
axes[1, 1].set_ylabel("Profit")
axes[1, 1].grid(True)


plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    "/Users/user/Downloads/matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv"
)


# ==========================================
# CLEAN REQUIRED COLUMNS
# ==========================================

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Revenue"] = pd.to_numeric(
    df["Revenue"],
    errors="coerce"
)

df["Profit"] = pd.to_numeric(
    df["Profit"],
    errors="coerce"
)


# ==========================================
# PREPARE DATA
# ==========================================

# Monthly Revenue
monthly_revenue = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Revenue"].sum()


# Category Revenue
category_revenue = df.groupby(
    "Product Category"
)["Revenue"].sum()


# Revenue Mean and Median
mean_revenue = df["Revenue"].mean()
median_revenue = df["Revenue"].median()


# ==========================================
# CREATE 2 x 3 DASHBOARD
# ==========================================

fig, axes = plt.subplots(
    2,
    3,
    figsize=(18, 10)
)


# ==========================================
# CHART 1: MONTHLY REVENUE
# ==========================================

axes[0, 0].plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o",
    linestyle="-"
)

axes[0, 0].set_title("Monthly Revenue")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue")

axes[0, 0].grid(True)

axes[0, 0].tick_params(
    axis="x",
    rotation=45
)


# ==========================================
# CHART 2: CATEGORY REVENUE
# ==========================================

bars = axes[0, 1].bar(
    category_revenue.index,
    category_revenue.values
)

axes[0, 1].set_title("Revenue by Product Category")
axes[0, 1].set_xlabel("Product Category")
axes[0, 1].set_ylabel("Revenue")

axes[0, 1].tick_params(
    axis="x",
    rotation=45
)


# Value labels
for bar in bars:

    height = bar.get_height()

    axes[0, 1].text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:.0f}",
        ha="center",
        va="bottom"
    )


# ==========================================
# CHART 3: REVENUE DISTRIBUTION
# ==========================================

axes[0, 2].hist(
    df["Revenue"].dropna(),
    bins=20,
    alpha=0.7,
    label="Revenue"
)

axes[0, 2].axvline(
    mean_revenue,
    linestyle="--",
    label=f"Mean = {mean_revenue:.0f}"
)

axes[0, 2].axvline(
    median_revenue,
    linestyle=":",
    label=f"Median = {median_revenue:.0f}"
)

axes[0, 2].set_title("Revenue Distribution")
axes[0, 2].set_xlabel("Revenue")
axes[0, 2].set_ylabel("Frequency")

axes[0, 2].grid(True)
axes[0, 2].legend()


# ==========================================
# CHART 4: REVENUE VS PROFIT
# ==========================================

axes[1, 0].scatter(
    df["Revenue"],
    df["Profit"],
    s=50,
    alpha=0.6,
    marker="o"
)

axes[1, 0].set_title("Revenue vs Profit")
axes[1, 0].set_xlabel("Revenue")
axes[1, 0].set_ylabel("Profit")

axes[1, 0].grid(True)


# ==========================================
# CHART 5: CATEGORY REVENUE SHARE
# ==========================================

axes[1, 1].pie(
    category_revenue.values,
    labels=category_revenue.index,
    autopct="%1.1f%%",
    startangle=90
)

axes[1, 1].set_title("Category Revenue Share")


# ==========================================
# EMPTY SIXTH SUBPLOT
# ==========================================

axes[1, 2].axis("off")


# ==========================================
# FINAL LAYOUT
# ==========================================

plt.tight_layout()

plt.show()
```
