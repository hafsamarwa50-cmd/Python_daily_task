# A.TITANIC DATASET

```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
tips = sns.load_dataset("tips")
```

```python
sns.catplot(
    data=titanic,
    x="sex",
    hue="survived",
    kind="count"
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    kind="count"
)

plt.title("Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    hue="survived",
    kind="count"
)

plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="sex",
    y="survived",
    kind="point",
    errorbar=None
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="sex",
    y="age",
    kind="violin"
)

plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="age",
    kind="box"
)

plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    kind="box",
    palette="Set2"
)

plt.title("Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="sex",
    y="fare",
    kind="violin",
    palette="Set2"
)

plt.title("Fare Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Fare")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="age",
    hue="sex",
    kind="box"
)

plt.title("Age Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="survived",
    hue="sex",
    kind="bar"
)

plt.title("Survival by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="embark_town",
    hue="survived",
    kind="count"
)

plt.title("Passenger Count by Embark Town and Survival")
plt.xlabel("Embark Town")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    kind="box"
)

plt.title("Fare by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
titanic["age_group"] = titanic["age"].apply(
    lambda age: "Child" if age < 18
    else ("Adult" if age < 60 else "Senior")
)
```

```python
sns.catplot(
    data=titanic,
    x="age_group",
    y="survived",
    kind="bar"
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="age",
    hue="sex",
    col="survived",
    kind="box"
)

plt.suptitle("Class, Gender and Survival", y=1.05)
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

```python
g = sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="box",
    palette="Set2",
    height=5,
    aspect=1.2
)

g.set_axis_labels("Passenger Class", "Fare")
g.set_titles("Survived = {col_name}")
g.fig.suptitle("Titanic Fare by Class, Gender and Survival", y=1.05)

plt.show()
```
