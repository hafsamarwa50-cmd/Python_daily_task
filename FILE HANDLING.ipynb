# ADVANCED FILE HANDLING:

```python
name = input("Name: ")
phone = input("Phone: ")

file = open("phonebook.txt", "a")
file.write(name + "," + phone + "\n")
file.close()

print("Saved.")
```

```python
file = open("log.txt", "r")

count = 0

for line in file:
    if "ERROR" in line:
        count += 1

print("Errors:", count)

file.close()
```

```python
file = open("notes.txt", "w")

while True:
    line = input()

    if line == "STOP":
        break

    file.write(line + "\n")

file.close()

print("Saved.")
```

```python
file = open("sample.txt", "r")
text = file.read()
file.close()

encrypted = ""

for ch in text:
    encrypted += chr(ord(ch) + 1)

file = open("encrypted.txt", "w")
file.write(encrypted)
file.close()

print("Encrypted.")
```

```python
expense = float(input("Enter expense: "))

file = open("expenses.txt", "a")
file.write(str(expense) + "\n")
file.close()

file = open("expenses.txt", "r")

total = 0

for line in file:
    total += float(line)

print("Total Expense =", total)

file.close()
```

```python
file = open("sample.txt", "r")
lines = file.readlines()
file.close()

file = open("clean.txt", "w")

for line in lines:
    if line.strip() != "":
        file.write(line)

file.close()

print("Blank lines removed.")
```

```python
file = open("sample.txt", "r")
lines = file.readlines()
file.close()

unique = set(lines)

file = open("unique.txt", "w")

for line in unique:
    file.write(line)

file.close()

print("Duplicates removed.")
```

```python
while True:

    print("\n1. Write")
    print("2. Read")
    print("3. Append")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        text = input("Enter text: ")
        file = open("menu.txt", "w")
        file.write(text)
        file.close()

    elif choice == "2":
        file = open("menu.txt", "r")
        print(file.read())
        file.close()

    elif choice == "3":
        text = input("Enter text: ")
        file = open("menu.txt", "a")
        file.write("\n" + text)
        file.close()

    elif choice == "4":
        print("Program Ended")
        break

    else:
        print("Invalid Choice")
```
