```python
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Equilateral Triangle")
    elif a == b or b == c or a == c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
else:
    print("Invalid Triangle")
```

```python
s1 = int(input("Enter Subject 1 Marks: "))
s2 = int(input("Enter Subject 2 Marks: "))
s3 = int(input("Enter Subject 3 Marks: "))
s4 = int(input("Enter Subject 4 Marks: "))
s5 = int(input("Enter Subject 5 Marks: "))

percentage = (s1 + s2 + s3 + s4 + s5) / 5

print("Percentage =", percentage)

if percentage >= 90:
    print("Grade: A+")
elif percentage >= 80:
    print("Grade: A")
elif percentage >= 70:
    print("Grade: B")
elif percentage >= 60:
    print("Grade: C")
else:
    print("Fail")
```

```python
num = int(input("Enter a number: "))

if num % 3 == 0 and num % 5 == 0:
    print("FizzBuzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
else:
    print(num)
```

```python
age = int(input("Enter age: "))
gender = input("Enter gender: ")

if age >= 18 and gender.lower() == "male":
    print("Eligible for Army")
elif age >= 18 and gender.lower() == "female":
    print("Eligible for Other Services")
else:
    print("Not Eligible")
```

```python
ch = input("Enter a character: ")

if ch.isupper():
    print("Uppercase Letter")
elif ch.islower():
    print("Lowercase Letter")
elif ch.isdigit():
    print("Digit")
else:
    print("Special Character")
```
