```python
class A:
    def method_a(self):
        print("Method from A")

class B(A):
    def method_b(self):
        print("Method from B")

class C:
    def method_c(self):
        print("Method from C")

class D(B, C):
    def method_d(self):
        print("Method from D")

obj = D()

obj.method_a()
obj.method_b()
obj.method_c()
obj.method_d()
```

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Employee.__init__(self, name, salary)
        self.marks = marks
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Marks:", self.marks)
        print("Department:", self.department)

ta = TeachingAssistant("Hafsa", 30000, 95, "IT")

ta.display()
```

```python
class Account:
    def __init__(self, account_no):
        self.account_no = account_no

class Savings(Account):
    def __init__(self, account_no, balance):
        super().__init__(account_no)
        self.balance = balance

class Loan:
    def __init__(self, loan_amount):
        self.loan_amount = loan_amount

class SavingsLoanAccount(Savings, Loan):
    def __init__(self, account_no, balance, loan_amount):
        Savings.__init__(self, account_no, balance)
        Loan.__init__(self, loan_amount)

    def display(self):
        print("Account No:", self.account_no)
        print("Balance:", self.balance)
        print("Loan Amount:", self.loan_amount)

obj = SavingsLoanAccount(101, 50000, 100000)

obj.display()
```

```python
class FileLogger:
    def log(self, message):
        print("File Log:", message)

class ConsoleLogger:
    def log(self, message):
        print("Console Log:", message)

def write_log(logger, message):
    logger.log(message)

f = FileLogger()
c = ConsoleLogger()

write_log(f, "Program Started")
write_log(c, "Program Completed")
```

```python
class Person:
    pass

class Employee(Person):
    pass

class Manager(Employee):
    pass

class HR(Employee):
    pass

class TeamLead(Manager, HR):
    def work(self):
        print("TeamLead manages the team")

class Freelancer:
    def work(self):
        print("Freelancer works independently")

def do_work(obj):
    obj.work()

t = TeamLead()
f = Freelancer()

do_work(t)
do_work(f)
```
