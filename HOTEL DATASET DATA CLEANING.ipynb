```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/user/Downloads/Hotel_booking_dataset - Hotel_booking_dataset.csv")

def clean_room_rate(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value.endswith("k"):
        value = value.replace("k", "")
        return float(value) * 1000

    value = (
        value.replace("$", "")
        .replace(",", "")
        .replace("inr", "")
        .strip()
    )

    return pd.to_numeric(value, errors="coerce")


df["Room Rate"] = df["Room Rate"].apply(clean_room_rate)

df.loc[
    df["Room Rate"] > 100000,
    "Room Rate"
] = np.nan

print(df["Room Rate"].head())
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/user/Downloads/Hotel_booking_dataset - Hotel_booking_dataset.csv")

df["Discount %"] = (
    df["Discount %"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "none": "0",
        "n/a": np.nan,
        "na": np.nan
    })
    .str.replace("%", "", regex=False)
)

df["Discount %"] = pd.to_numeric(
    df["Discount %"],
    errors="coerce"
)

df.loc[
    (df["Discount %"] < 0) |
    (df["Discount %"] > 100),
    "Discount %"
] = np.nan

print(df["Discount %"].head())
```
