```python
import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. Load dataset
# --------------------------------------------------

df = pd.read_csv(
    "/Users/user/Downloads/messy_customer_loan_dataset - messy_customer_loan_dataset.csv"
)

print("Original shape:", df.shape)

# --------------------------------------------------
# 2. Clean column names
# --------------------------------------------------

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace("-", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("?", "", regex=False)
)

print("\nCleaned column names:")
print(df.columns.tolist())

# --------------------------------------------------
# 3. Remove duplicate rows
# --------------------------------------------------

duplicates = df.duplicated().sum()

df = df.drop_duplicates()

print("\nDuplicate rows removed:", duplicates)

# --------------------------------------------------
# 4. Clean Age
# --------------------------------------------------

df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)

# Invalid ages
df.loc[
    (df["age"] < 0) | (df["age"] > 100),
    "age"
] = np.nan

# --------------------------------------------------
# 5. Clean Annual Income
# --------------------------------------------------

def clean_income(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value in ["n/a", "na", "nan", "-", "unknown"]:
        return np.nan

    try:
        if "l" in value:
            value = value.replace("l", "")
            return float(value) * 100000

        value = (
            value
            .replace("₹", "")
            .replace("inr", "")
            .replace(",", "")
            .strip()
        )

        return float(value)

    except:
        return np.nan


df["annual_income"] = df["annual_income"].apply(clean_income)

# --------------------------------------------------
# 6. Clean Credit Score
# --------------------------------------------------

df["credit_score"] = pd.to_numeric(
    df["credit_score"],
    errors="coerce"
)

# Valid credit score range: 300–850
df.loc[
    (df["credit_score"] < 300) |
    (df["credit_score"] > 850),
    "credit_score"
] = np.nan

# --------------------------------------------------
# 7. Clean Loan Amount
# --------------------------------------------------

df["loan_amount_inr"] = (
    df["loan_amount_inr"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("INR", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["loan_amount_inr"] = pd.to_numeric(
    df["loan_amount_inr"],
    errors="coerce"
)

# --------------------------------------------------
# 8. Clean Monthly Visits
# --------------------------------------------------

def clean_visits(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    word_numbers = {
        "zero": 0,
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5,
        "six": 6,
        "seven": 7,
        "eight": 8,
        "nine": 9,
        "ten": 10
    }

    if value in word_numbers:
        return word_numbers[value]

    if value in ["n/a", "na", "-", "unknown"]:
        return np.nan

    try:
        return float(value)
    except:
        return np.nan


df["monthly_visits"] = df["monthly_visits"].apply(clean_visits)

# --------------------------------------------------
# 9. Clean Debt to Income
# --------------------------------------------------

def clean_dti(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    try:

        if value.endswith("%"):
            return float(value[:-1]) / 100

        elif value.endswith("x"):
            return float(value[:-1])

        else:
            return float(value)

    except:
        return np.nan


df["debt_to_income"] = df["debt_to_income"].apply(clean_dti)

# Invalid DTI values
df.loc[
    (df["debt_to_income"] < 0) |
    (df["debt_to_income"] > 1),
    "debt_to_income"
] = np.nan

# --------------------------------------------------
# 10. Clean Employment Type
# --------------------------------------------------

df["employment_type"] = (
    df["employment_type"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.title()
)

# --------------------------------------------------
# 11. Clean Education Level
# --------------------------------------------------

df["education_level"] = (
    df["education_level"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.title()
)

# --------------------------------------------------
# 12. Clean City
# --------------------------------------------------

df["city"] = (
    df["city"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["city"] = df["city"].replace({
    "hyderbad": "Hyderabad",
    "hyd": "Hyderabad",
    "hyderabad": "Hyderabad",
    "bangalore": "Bengaluru",
    "bengaluru": "Bengaluru"
})

df["city"] = df["city"].str.title()

# --------------------------------------------------
# 13. Clean Customer Segment
# --------------------------------------------------

df["customer_segment"] = (
    df["customer_segment"]
    .astype(str)
    .str.strip()
    .str.lower()
    .str.title()
)

# --------------------------------------------------
# 14. Clean Loan Type
# --------------------------------------------------

df["loan_type"] = (
    df["loan_type"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["loan_type"] = df["loan_type"].replace({
    "personal": "Personal Loan",
    "personal loan": "Personal Loan",
    "home": "Home Loan",
    "home loan": "Home Loan",
    "car": "Car Loan",
    "car loan": "Car Loan",
    "education": "Education Loan",
    "education loan": "Education Loan"
})

# --------------------------------------------------
# 15. Clean Loan Approved
# --------------------------------------------------

df["loan_approved"] = (
    df["loan_approved"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["loan_approved"] = df["loan_approved"].replace({
    "yes": "Yes",
    "y": "Yes",
    "1": "Yes",
    "no": "No",
    "n": "No",
    "0": "No"
})

# --------------------------------------------------
# 16. Handle missing numerical values
# --------------------------------------------------

numeric_columns = [
    "age",
    "annual_income",
    "credit_score",
    "loan_amount_inr",
    "monthly_visits",
    "debt_to_income"
]

for column in numeric_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )

# --------------------------------------------------
# 17. Detect and treat Loan Amount outliers
# --------------------------------------------------

Q1 = df["loan_amount_inr"].quantile(0.25)
Q3 = df["loan_amount_inr"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df["loan_amount_inr"] = df["loan_amount_inr"].clip(
    lower=lower_limit,
    upper=upper_limit
)

# --------------------------------------------------
# 18. Detect and treat Annual Income outliers
# --------------------------------------------------

Q1 = df["annual_income"].quantile(0.25)
Q3 = df["annual_income"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df["annual_income"] = df["annual_income"].clip(
    lower=lower_limit,
    upper=upper_limit
)

# --------------------------------------------------
# 19. Final Data Quality Check
# --------------------------------------------------

print("\n========== FINAL DATA QUALITY CHECK ==========")

print("\nRemaining missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nFinal shape:")
print(df.shape)

print("\nFirst 10 rows of cleaned dataset:")
print(df.head(10))
```
