# HIGH LEVEL:

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

result = []

for word in words:
    if word[0].lower() not in "aeiou":
        result.append(word)

print(" ".join(result))
```

```python
text = input("Enter a sentence: ")

text = text.replace(" ", "-")

words = text.split("-")

for word in words:
    if len(word) > 4:
        print(word)
```

```python
students = ["Rahul", "Aman", "Neha", "Priya", "Kiran"]

joined = ", ".join(students)

print("Original:", joined)

students[0] = "New Student"

print("Updated:", ", ".join(students))
```

```python
paragraph = input("Enter a paragraph: ")

sentences = paragraph.split(".")

clean = []

for sentence in sentences:
    sentence = sentence.strip()
    if sentence != "":
        clean.append(sentence)

print("- " + "\n- ".join(clean))
```

```python
text = input("Enter a sentence: ")

text = text.strip()

text = text.replace(",", " ")

words = text.split()

result = " ".join(words)

print(result)
```
