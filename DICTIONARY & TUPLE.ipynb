# MEDIUM LEVEL:

```python
d = {"a": 1, "b": 2}

d.update({"c": 3, "d": 4})

print(d)
```

```python
student = {
    "name": "Rahul",
    "age": 22,
    "city": "Delhi"
}

student.pop("age")

print(student)
```

```python
d = {"x": 10, "y": 20, "z": 30}

item = d.popitem()

print("Removed:", item)
print(d)
```

```python
original = {"a": 1, "b": 2}

copy_dict = original.copy()

print(copy_dict)
```

```python
colors = ["red", "green", "blue"]

d = dict.fromkeys(colors, 0)

print(d)
```

```python
tup = (10, "Hello", 3.14, True)

print(tup)

tup[1] = "Python"
```

```python
d = {"name": "Sumit"}

d.setdefault("age", 25)

print(d)
```

# HARD/ MIXED LEVEL:

```python
d = {"a": 1}

d.update({
    "b": 2,
    "c": 3,
    "d": 4
})

print(d)

removed = d.popitem()

print("Removed:", removed)
print(d)
```

```python
tup = (10, 20, 30, 20, 40, 50)

print("Index of 20:", tup.index(20))
print("Count of 20:", tup.count(20))
```

```python
students = {}

for i in range(3):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks

print(students)
```

```python
data = {
    "a": 10,
    "b": 20,
    "c": 10
}

for key, value in data.items():
    if value == 10:
        print(key)
```

```python
d = {
    "a": 1,
    "b": 2,
    "c": 3,
    "d": 4,
    "e": 5
}

copy_d = d.copy()

d.clear()

print("Original:", d)
print("Copy:", copy_d)
```

```python
student = {
    "name": "Rahul",
    "age": 20,
    "city": "Delhi",
    "course": "Python"
}

student.update({
    "college": "ABC College",
    "marks": 95
})

student.popitem()

print(student.items())
```

```python
tup = (25, 10, 30, 40, 25, 60, 70, 80)

if 25 in tup:
    print("Index:", tup.index(25))
    print("Count:", tup.count(25))

lst = list(tup)

lst.sort(reverse=True)

print(lst)
```

```python
students = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks

highest = max(students, key=students.get)

print("Top Student:", highest)
print("Marks:", students[highest])
```
