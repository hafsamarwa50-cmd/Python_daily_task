# METHOD OVERLOADING:

```python
class Calculator:
    def add(self, a, b, c=None):
        if c is None:
            return a + b
        else:
            return a + b + c


obj = Calculator()

print(obj.add(10, 20))
print(obj.add(10, 20, 30))
```

```python
class Area:
    def calculate(self, length, breadth=None):
        if breadth is None:
            return length * length
        else:
            return length * breadth


obj = Area()

print("Square Area =", obj.calculate(5))
print("Rectangle Area =", obj.calculate(5, 8))
```

```python
class Student:
    def details(self, name, age=None, city=None):
        if age is None:
            print("Name:", name)
        elif city is None:
            print("Name:", name)
            print("Age:", age)
        else:
            print("Name:", name)
            print("Age:", age)
            print("City:", city)


obj = Student()

obj.details("Hafsa")
print()

obj.details("Hafsa", 20)
print()

obj.details("Hafsa", 20, "Hyderabad")
```

```python
class MathOperations:
    def multiply(self, *args):
        result = 1

        for i in args:
            result = result * i

        return result


obj = MathOperations()

print(obj.multiply(2, 3))
print(obj.multiply(2, 3, 4))
print(obj.multiply(2, 3, 4, 5))
print(obj.multiply(1, 2, 3, 4, 5, 6))
```

```python
class Display:
    def show(self, value):
        if isinstance(value, int):
            print("Integer Value:", value)

        elif isinstance(value, str):
            print("String Value:", value)

        elif isinstance(value, list):
            print("List Value:", value)

        else:
            print("Unknown Type")


obj = Display()

obj.show(100)
obj.show("Python")
obj.show([10, 20, 30])
```

```python
class Volume:
    def calculate(self, length, breadth=None, height=None):
        if breadth is None and height is None:
            return length ** 3
        else:
            return length * breadth * height


obj = Volume()

print("Cube Volume =", obj.calculate(4))
print("Cuboid Volume =", obj.calculate(4, 5, 6))
```

```python
class PrintData:
    def print_data(self, *args):
        for value in args:
            print(value)


obj = PrintData()

obj.print_data("Python")
print()

obj.print_data("Python", 100, "AI", 3.14)
```

# METHOD OVERRIDING:

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
class Vehicle:
    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    def start(self):
        print("Car starts with a key")


class Bike(Vehicle):
    def start(self):
        print("Bike starts with a self-start button")


car = Car()
bike = Bike()

car.start()
bike.start()
```

```python
class Employee:
    def __init__(self, salary):
        self.salary = salary

    def calculate_salary(self):
        return self.salary


class Manager(Employee):
    def __init__(self, salary, bonus):
        super().__init__(salary)
        self.bonus = bonus

    def calculate_salary(self):
        return self.salary + self.bonus


manager = Manager(50000, 10000)

print("Manager Salary =", manager.calculate_salary())
```

```python
class Shape:
    def area(self):
        print("Area of shape")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Circle Area =", 3.14 * self.radius * self.radius)


class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        print("Rectangle Area =", self.length * self.breadth)


circle = Circle(5)
rectangle = Rectangle(4, 6)

circle.area()
rectangle.area()
```

```python
class BankAccount:
    def interest(self):
        print("General Interest Rate")


class SavingsAccount(BankAccount):
    def interest(self):
        print("Savings Account Interest: 6%")


class CurrentAccount(BankAccount):
    def interest(self):
        print("Current Account Interest: 2%")


savings = SavingsAccount()
current = CurrentAccount()

savings.interest()
current.interest()
```

```python
class Parent:
    def display(self):
        print("This is Parent class")


class Child(Parent):
    def display(self):
        super().display()
        print("This is Child class")


obj = Child()
obj.display()
```

```python
class Bird:
    def fly(self):
        print("Bird can fly")


class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly")


penguin = Penguin()
penguin.fly()
```

# OPERATOR OVERLOADING:

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)


p1 = Point(2, 3)
p2 = Point(4, 5)

result = p1 + p2

print("X =", result.x)
print("Y =", result.y)
```

```python
class Book:
    def __init__(self, pages):
        self.pages = pages

    def __add__(self, other):
        return self.pages + other.pages


book1 = Book(250)
book2 = Book(300)

print("Total Pages =", book1 + book2)
```

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __gt__(self, other):
        return self.marks > other.marks


student1 = Student(85)
student2 = Student(75)

print(student1 > student2)
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def __eq__(self, other):
        return self.area() == other.area()


r1 = Rectangle(4, 5)
r2 = Rectangle(2, 10)

print(r1 == r2)
```

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __sub__(self, other):
        return Money(self.amount - other.amount)


m1 = Money(1000)
m2 = Money(400)

add_result = m1 + m2
sub_result = m1 - m2

print("Addition =", add_result.amount)
print("Subtraction =", sub_result.amount)
```

```python
class Distance:
    def __init__(self, distance):
        self.distance = distance

    def __lt__(self, other):
        return self.distance < other.distance

    def __gt__(self, other):
        return self.distance > other.distance

    def __eq__(self, other):
        return self.distance == other.distance


d1 = Distance(25)
d2 = Distance(30)

print(d1 < d2)
print(d1 > d2)
print(d1 == d2)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


n1 = Number(10)
n2 = Number(20)

result = n1 + n2

print(result.value)
```
