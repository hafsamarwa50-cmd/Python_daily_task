```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number =", reverse)
```

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
num = int(input("Enter a number: "))

original = num
total = 0

while num > 0:
    digit = num % 10
    total = total + digit ** 3
    num = num // 10

if total == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
n = int(input("Enter how many terms: "))

a = 0
b = 1
count = 1

while count <= n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```

```python
total = 0

while True:
    num = int(input("Enter a number (0 to stop): "))

    if num == 0:
        break

    total += num

print("Total Sum =", total)
```

```python
while True:
    year = input("Enter year (or type stop): ")

    if year.lower() == "stop":
        break

    year = int(year)

    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print("Leap Year")
    else:
        print("Not a Leap Year")
```

```python
num = 2

while num <= 100:
    i = 2
    prime = True

    while i < num:
        if num % i == 0:
            prime = False
            break
        i += 1

    if prime:
        print(num)

    num += 1
```

```python
total = 0
count = 0

while True:
    marks = float(input("Enter marks (negative to stop): "))

    if marks < 0:
        break

    total += marks
    count += 1

if count > 0:
    average = total / count
    print("Average =", average)
else:
    print("No marks entered.")
```

```python
correct_username = "admin"
correct_password = "python123"

attempts = 3

while attempts > 0:
    username = input("Enter Username: ")
    password = input("Enter Password: ")

    if username == correct_username and password == correct_password:
        print("Login Successful")
        break

    else:
        attempts -= 1
        print("Wrong Username or Password")
        print("Attempts Left:", attempts)

if attempts == 0:
    print("Account Blocked")
```
