```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

class Bike(Vehicle):
    def __init__(self, brand, engine_cc):
        super().__init__(brand)
        self.engine_cc = engine_cc

c = Car("Toyota", 4)
b = Bike("Honda", 150)

print("Car:", c.brand, c.doors)
print("Bike:", b.brand, b.engine_cc)
```

```python
class Shape:
    def area(self):
        print("Area not defined.")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Circle Area =", 3.14 * self.radius * self.radius)

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        print("Square Area =", self.side * self.side)

c = Circle(5)
s = Square(4)

c.area()
s.area()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print("Student")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def display(self):
        print("\nTeacher")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Subject:", self.subject)

class Staff(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department

    def display(self):
        print("\nStaff")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Department:", self.department)

s = Student("Hafsa", 20, 95)
t = Teacher("Ahmed", 40, "Python")
st = Staff("Sara", 35, "Accounts")

s.display()
t.display()
st.display()
```

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

class Savings(Account):
    def show(self):
        print("Savings Balance:", self.balance)
        print("Interest: 5%")

class Current(Account):
    def show(self):
        print("Current Balance:", self.balance)
        print("Overdraft: £1000")

class FixedDeposit(Account):
    def show(self):
        print("FD Balance:", self.balance)
        print("Maturity: 5 Years")

s = Savings(10000)
c = Current(20000)
f = FixedDeposit(50000)

s.show()
c.show()
f.show()
```

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def calculate_salary(self):
        pass

class FullTime(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        print(self.name, "Salary =", self.monthly_salary)

class PartTime(Employee):
    def __init__(self, name, hours, rate):
        super().__init__(name)
        self.hours = hours
        self.rate = rate

    def calculate_salary(self):
        print(self.name, "Salary =", self.hours * self.rate)

f = FullTime("Rahul", 50000)
p = PartTime("Asha", 40, 500)

f.calculate_salary()
p.calculate_salary()
```

```python
class University:
    def __init__(self, university_name):
        self.university_name = university_name

    def display_university(self):
        print("University:", self.university_name)

class Student(University):
    def study(self):
        print("Student is studying")

class Professor(University):
    def teach(self):
        print("Professor is teaching")

class Librarian(University):
    def manage_books(self):
        print("Librarian manages books")

s = Student("ABC University")
p = Professor("ABC University")
l = Librarian("ABC University")

s.display_university()
s.study()

print()

p.display_university()
p.teach()

print()

l.display_university()
l.manage_books()
```
