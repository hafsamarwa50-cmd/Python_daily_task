# Q29

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(
    "/Users/user/Downloads/customer_data - customer_data.csv"
)

# 1. Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# 2. Remove duplicate customer IDs
df = df.drop_duplicates(
    subset="customer_id",
    keep="first"
)

# 3. Clean customer_segment
df["customer_segment"] = (
    df["customer_segment"]
    .astype("string")
    .str.strip()
    .str.title()
)

# 4. Clean gender
df["gender"] = (
    df["gender"]
    .astype("string")
    .str.strip()
    .str.lower()
)

df["gender"] = df["gender"].replace({
    "m": "Male",
    "male": "Male",
    "man": "Male",
    "f": "Female",
    "female": "Female",
    "woman": "Female"
})

# 5. Clean is_active
df["is_active"] = (
    df["is_active"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map({
        "yes": 1,
        "y": 1,
        "1": 1,
        "no": 0,
        "n": 0,
        "0": 0
    })
)

df["is_active"] = pd.to_numeric(
    df["is_active"],
    errors="coerce"
)

print(df["is_active"].unique())

df["is_active"] = pd.to_numeric(
    df["is_active"],
    errors="coerce"
)

# 6. Clean churn_flag
df["churn_flag"] = (
    df["churn_flag"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map({
        "yes": 1,
        "no": 0
    })
)

df["churn_flag"] = pd.to_numeric(
    df["churn_flag"],
    errors="coerce"
)

print(df["churn_flag"].unique())

# 7. Clean income
df["income_usd"] = (
    df["income_usd"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace("USD", "", regex=False)
    .str.strip()
)

df["income_usd"] = pd.to_numeric(
    df["income_usd"],
    errors="coerce"
)

# 8. Clean age
df["age"] = (
    df["age"]
    .astype("string")
    .str.extract(r"(\d+)", expand=False)
)

df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)

# 9. Replace unrealistic ages
df.loc[
    (df["age"] < 18) | (df["age"] > 100),
    "age"
] = np.nan

# 10. Clean credit score
df["credit_score"] = (
    df["credit_score"]
    .astype("string")
    .str.replace("pts", "", case=False, regex=False)
    .str.strip()
)

df["credit_score"] = pd.to_numeric(
    df["credit_score"],
    errors="coerce"
)

# 11. Clean tenure_years
df["tenure_years"] = pd.to_numeric(
    df["tenure_years"],
    errors="coerce"
)

df["tenure_years"] = df["tenure_years"].replace(
    [-999, 99999],
    np.nan
)

# 12. Clean satisfaction
df["satisfaction"] = pd.to_numeric(
    df["satisfaction"],
    errors="coerce"
)

df["satisfaction"] = df["satisfaction"].replace(
    [-999, 99999],
    np.nan
)

# 13. Clean signup date
df["signup_date"] = pd.to_datetime(
    df["signup_date"],
    errors="coerce"
)

# 14. Clean lead source
df["lead_source"] = (
    df["lead_source"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.replace("_", " ", regex=False)
)

df["lead_source"] = df["lead_source"].replace({
    "web": "Website",
    "website": "Website",
    "referral": "Referral",
    "google ads": "Google Ads"
})

# 15. Detect and replace income outliers
Q1 = df["income_usd"].quantile(0.25)
Q3 = df["income_usd"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df.loc[
    (df["income_usd"] < lower) |
    (df["income_usd"] > upper),
    "income_usd"
] = np.nan

# 16. Detect and replace credit score outliers
Q1 = df["credit_score"].quantile(0.25)
Q3 = df["credit_score"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df.loc[
    (df["credit_score"] < lower) |
    (df["credit_score"] > upper),
    "credit_score"
] = np.nan

# 17. Fill numerical missing values with median
numeric_columns = df.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:
    df[col] = df[col].fillna(
        df[col].median()
    )

# 18. Fill categorical missing values with mode
categorical_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(
            df[col].mode()[0]
        )

# Final cleaned DataFrame
cleaned_df = df.copy()

print("Final cleaned DataFrame:")
print(cleaned_df.head())

print("\nFinal shape:")
print(cleaned_df.shape)
```
