```python
class Father:
    def father_info(self):
        print("Father: John")


class Mother:
    def mother_info(self):
        print("Mother: Mary")


class Child(Father, Mother):
    def child_info(self):
        print("Child: Alex")


c = Child()
c.father_info()
c.mother_info()
c.child_info()
```

```python
class Teacher:
    def __init__(self, subject):
        self.subject = subject


class Author:
    def __init__(self, book):
        self.book = book


class Person(Teacher, Author):
    def __init__(self, subject, book):
        Teacher.__init__(self, subject)
        Author.__init__(self, book)

    def display(self):
        print("Subject:", self.subject)
        print("Book:", self.book)


p = Person("Python", "Learning Python")
p.display()
```

# MULTIPLE INHERITANCE WITH STATIC METHODS:

```python
class Father:
    @staticmethod
    def father_method():
        print("Static Method from Father")


class Mother:
    @staticmethod
    def mother_method():
        print("Static Method from Mother")


class Child(Father, Mother):
    pass


Child.father_method()
Child.mother_method()
```

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b


class AdvancedMath(MathUtils):
    pass


result = AdvancedMath.add(10, 20)
print("Sum =", result)
```

```python
class Validator:
    @staticmethod
    def is_valid(name):
        return len(name) > 0


class Form(Validator):
    def submit(self, name):
        if self.is_valid(name):
            print("Form Submitted")
        else:
            print("Invalid Name")


f = Form()
f.submit("Hafsa")
```

```python
class Converter:
    @staticmethod
    def to_celsius(fahrenheit):
        return (fahrenheit - 32) * 5 / 9


class Weather:
    def show(self):
        print("Weather Report")


class WeatherApp(Converter, Weather):
    pass


app = WeatherApp()
app.show()

temp = WeatherApp.to_celsius(95)
print("Temperature:", temp, "°C")
```

```python
class Logger:
    @staticmethod
    def log(message):
        print("LOG:", message)


class User:
    def login(self):
        print("User Logged In")


class Application(Logger, User):
    pass


app = Application()
app.login()
Application.log("Application Started")
```

```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b


class Display:
    def show(self, value):
        print("Result =", value)


class MyApp(Calculator, Display):
    pass


obj = MyApp()
result = obj.add(15, 25)
obj.show(result)
```

# MULTIPLE INHERITANCE WITH BSTRACT METHOD:

```python
from abc import ABC, abstractmethod
```

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def draw(self):
        pass


class Color(ABC):
    @abstractmethod
    def fill(self):
        pass


class Rectangle(Shape, Color):
    def draw(self):
        print("Drawing Rectangle")

    def fill(self):
        print("Filling Rectangle with Blue Color")


obj = Rectangle()
obj.draw()
obj.fill()
```

```python
from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def make_payment(self):
        pass


class Notification(ABC):
    @abstractmethod
    def send_notification(self):
        pass


class OnlinePayment(Payment, Notification):
    def make_payment(self):
        print("Payment Successful")

    def send_notification(self):
        print("Notification Sent")


obj = OnlinePayment()
obj.make_payment()
obj.send_notification()
```

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass


class Electric(ABC):
    @abstractmethod
    def charge(self):
        pass


class ElectricCar(Vehicle, Electric):
    def start(self):
        print("Electric Car Started")

    def charge(self):
        print("Battery Charging")


car = ElectricCar()
car.start()
car.charge()
```

```python
from abc import ABC, abstractmethod


class Readable(ABC):
    @abstractmethod
    def read(self):
        pass


class Writable(ABC):
    @abstractmethod
    def write(self):
        pass


class File(Readable, Writable):
    def read(self):
        print("Reading File")

    def write(self):
        print("Writing File")


f = File()
f.read()
f.write()
```

```python
from abc import ABC, abstractmethod


class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass


class Swimmable(ABC):
    @abstractmethod
    def swim(self):
        pass


class Duck(Flyable, Swimmable):
    def fly(self):
        print("Duck is Flying")

    def swim(self):
        print("Duck is Swimming")


d = Duck()
d.fly()
d.swim()
```

```python
from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass


class Walk:
    def walk(self):
        print("Animal is Walking")


class Dog(Animal, Walk):
    def sound(self):
        print("Dog Barks")


d = Dog()
d.sound()
d.walk()
```

```python
from abc import ABC, abstractmethod


class A(ABC):
    @abstractmethod
    def show(self):
        pass


class B(ABC):
    @abstractmethod
    def show(self):
        pass


class C(A, B):
    def show(self):
        print("Implemented show() method")


obj = C()
obj.show()
```

# HARD LEVEL QUESTIONS:

```python
class Employee:
    def get_salary(self):
        return 50000


class Person:
    def get_details(self):
        return "Name: Hafsa"


class Manager(Employee, Person):
    def display(self):
        print(self.get_details())
        print("Salary:", self.get_salary())


m = Manager()
m.display()
```

```python
from abc import ABC, abstractmethod


class Device(ABC):
    @abstractmethod
    def power_on(self):
        pass


class Network:
    def connect(self):
        print("Connected to Wi-Fi")


class SmartPhone(Device, Network):
    def power_on(self):
        print("SmartPhone Powered On")


phone = SmartPhone()
phone.power_on()
phone.connect()
```
