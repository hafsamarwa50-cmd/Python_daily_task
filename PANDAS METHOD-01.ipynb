```python
class NegativeBalanceError(Exception):
    pass


class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Balance after deposit:", self.__balance)

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise NegativeBalanceError("Balance cannot be negative.")
        self.__balance -= amount
        print("Balance after withdrawal:", self.__balance)


b = BankAccount(5000)

try:
    b.deposit(2000)
    b.withdraw(8000)
except NegativeBalanceError as e:
    print(e)
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    def __init__(self, value):
        self.value = value

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        super().__init__(radius)

    def area(self):
        return 3.14 * self.value * self.value


class Square(Shape):

    def __init__(self, side):
        super().__init__(side)

    def area(self):
        return self.value * self.value


c = Circle(5)
s = Square(4)

print("Circle Area:", c.area())
print("Square Area:", s.area())
```

```python
class EmptyMarksError(Exception):
    pass


class Student:

    def __init__(self):
        self.__marks = []

    # Setter
    def set_marks(self, marks):
        for mark in marks:
            if mark < 0 or mark > 100:
                print("Invalid Marks! Marks should be between 0 and 100.")
                return
        self.__marks = marks

    # Getter
    def get_marks(self):
        return self.__marks

    # Calculate Average
    def average(self):
        if len(self.__marks) == 0:
            raise EmptyMarksError("Marks list is empty.")
        return sum(self.__marks) / len(self.__marks)

    # Operator Overloading
    def __gt__(self, other):
        return self.average() > other.average()


try:
    s1 = Student()
    s2 = Student()

    marks1 = list(map(int, input("Enter marks of Student 1 separated by space: ").split()))
    marks2 = list(map(int, input("Enter marks of Student 2 separated by space: ").split()))

    s1.set_marks(marks1)
    s2.set_marks(marks2)

    print("Student 1 Marks:", s1.get_marks())
    print("Student 2 Marks:", s2.get_marks())

    print("Student 1 Average:", s1.average())
    print("Student 2 Average:", s2.average())

    if s1 > s2:
        print("Student 1 has higher average.")
    else:
        print("Student 2 has higher average.")

except EmptyMarksError as e:
    print(e)
```

```python
class InsufficientBalanceError(Exception):
    pass


class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount Deposited Successfully.")

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient Balance.")
        self.balance -= amount
        print("Amount Withdrawn Successfully.")

    def show(self):
        print("Name:", self.name)
        print("Balance:", self.balance)

    def save(self):
        file = open("accounts.txt", "w")
        file.write(self.name + "\n")
        file.write(str(self.balance))
        file.close()


try:
    file = open("accounts.txt", "r")
    name = file.readline().strip()
    balance = float(file.readline())
    file.close()

    account = BankAccount(name, balance)

except FileNotFoundError:
    print("No account found. Create a new account.")

    name = input("Enter Account Holder Name: ")
    balance = float(input("Enter Opening Balance: "))

    account = BankAccount(name, balance)
    account.save()


while True:

    print("\n------ BANK MENU ------")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Show Account")
    print("4. Save & Exit")

    choice = input("Enter Choice: ")

    try:

        if choice == "1":
            amount = float(input("Enter Deposit Amount: "))
            account.deposit(amount)

        elif choice == "2":
            amount = float(input("Enter Withdraw Amount: "))
            account.withdraw(amount)

        elif choice == "3":
            account.show()

        elif choice == "4":
            account.save()
            print("Account Saved Successfully.")
            print("Thank You!")
            break

        else:
            print("Invalid Choice.")

    except ValueError:
        print("Please enter valid numeric values.")

    except InsufficientBalanceError as e:
        print(e)
```

```python
class StudentNotFoundError(Exception):
    pass


students = {}

while True:

    print("\n------ STUDENT MANAGEMENT SYSTEM ------")
    print("1. Add Student")
    print("2. Update Marks")
    print("3. Delete Student")
    print("4. Display All Students")
    print("5. Exit")

    choice = input("Enter your choice: ")

    try:

        if choice == "1":
            name = input("Enter Student Name: ")
            marks = int(input("Enter Marks: "))
            students[name] = marks
            print("Student Added Successfully.")

        elif choice == "2":
            name = input("Enter Student Name to Update: ")

            if name not in students:
                raise StudentNotFoundError("Student not found.")

            marks = int(input("Enter New Marks: "))
            students[name] = marks
            print("Marks Updated Successfully.")

        elif choice == "3":
            name = input("Enter Student Name to Delete: ")

            if name not in students:
                raise StudentNotFoundError("Student not found.")

            del students[name]
            print("Student Deleted Successfully.")

        elif choice == "4":

            if len(students) == 0:
                print("No Student Records Found.")
            else:
                print("\nStudent Records")
                for name, marks in students.items():
                    print(name, ":", marks)

        elif choice == "5":
            print("Program Ended.")
            break

        else:
            print("Invalid Choice.")

    except ValueError:
        print("Please enter valid marks.")

    except StudentNotFoundError as e:
        print(e)
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
        self.history = [salary]

    # Getter
    def get_salary(self):
        return self.__salary

    # Setter
    def set_salary(self, new_salary):
        max_salary = self.__salary + (self.__salary * 15 / 100)

        if new_salary > max_salary:
            print("Salary increase cannot exceed 15%.")
        else:
            self.__salary = new_salary
            self.history.append(new_salary)
            print("Salary Updated Successfully.")

    # Display Salary History
    def show_history(self):
        print("Salary History:", self.history)

    # Static Method
    @staticmethod
    def calculate_tax(salary):
        return salary * 0.10

    # Operator Overloading
    def __gt__(self, other):
        return self.__salary > other.__salary


emp1 = Employee("Rahul", 50000)
emp2 = Employee("Priya", 60000)

print("Employee 1 Salary:", emp1.get_salary())
print("Employee 2 Salary:", emp2.get_salary())

new_salary = float(input("Enter New Salary for Rahul: "))
emp1.set_salary(new_salary)

print("Updated Salary:", emp1.get_salary())

emp1.show_history()

print("Tax on Rahul's Salary:", Employee.calculate_tax(emp1.get_salary()))

if emp1 > emp2:
    print("Rahul has a higher salary.")
else:
    print("Priya has a higher salary.")
```

```python
from abc import ABC, abstractmethod


# Custom Exception
class InvalidSpeedError(Exception):
    pass


# Abstract Class
class Vehicle(ABC):

    def __init__(self, name, speed):
        self.name = name
        self.__speed = speed

    # Getter
    def get_speed(self):
        return self.__speed

    # Setter
    def set_speed(self, speed):
        if speed < 0:
            raise InvalidSpeedError("Speed cannot be negative.")
        self.__speed = speed

    # Operator Overloading
    def __gt__(self, other):
        return self.__speed > other.__speed

    @abstractmethod
    def details(self):
        pass


# Multiple Inheritance
class Electric:

    def battery(self):
        print("Battery Powered Vehicle")


class Fuel:

    def fuel(self):
        print("Fuel Powered Vehicle")


# Child Class
class Car(Vehicle, Electric, Fuel):

    def __init__(self, name, speed):
        super().__init__(name, speed)

    def details(self):
        print("Vehicle Name:", self.name)
        print("Speed:", self.get_speed(), "km/h")

    # Duck Typing Method
    def start(self):
        print(self.name, "Started")


# Another Class for Duck Typing
class Bike:

    def start(self):
        print("Bike Started")


try:

    name = input("Enter Vehicle Name: ")
    speed = int(input("Enter Speed: "))

    car = Car(name, speed)

    car.details()
    car.battery()
    car.fuel()

    # Save to File
    file = open("vehicle.txt", "w")
    file.write("Vehicle Name: " + car.name + "\n")
    file.write("Speed: " + str(car.get_speed()))
    file.close()

    print("Vehicle Details Saved Successfully.")

    # Operator Overloading
    car2 = Car("BMW", 180)

    if car > car2:
        print(car.name, "is Faster")
    else:
        print(car2.name, "is Faster")

    # Duck Typing
    bike = Bike()

    objects = [car, bike]

    for obj in objects:
        obj.start()

except InvalidSpeedError as e:
    print(e)
```
