# RAISING EXCEPTIONS AND CUSTOM EXCEPTIONS:

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or above.")

    print("Eligible")


try:
    check_age(16)

except ValueError as e:
    print(e)
```

```python
class NegativeNumberError(Exception):
    pass


try:
    num = int(input("Enter number: "))

    if num < 0:
        raise NegativeNumberError("Negative Number Entered.")

    print(num)

except NegativeNumberError as e:
    print(e)
```

```python
class InsufficientBalanceError(Exception):
    pass


balance = 5000

try:
    amount = int(input("Enter withdrawal amount: "))

    if amount > balance:
        raise InsufficientBalanceError("Insufficient Balance")

    balance -= amount

    print("Remaining Balance =", balance)

except InsufficientBalanceError as e:
    print(e)
```

```python
data = 100

try:
    if not isinstance(data, str):
        raise TypeError("Input must be a string.")

except TypeError as e:
    print(e)
```

```python
class PasswordError(Exception):
    pass


try:
    password = input("Enter Password: ")

    if len(password) < 8:
        raise PasswordError("Password must contain at least 8 characters.")

    print("Valid Password")

except PasswordError as e:
    print(e)
```

```python
try:
    raise Exception("This is a custom error message.")

except Exception as e:
    print(e)
```

```python
try:
    raise ValueError("Forced Exception")

except ValueError as e:
    print(e)
```

# HARD QUESTION:

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    op = input("Enter operator (+,-,*,/): ")

    if op == "+":
        print(a + b)

    elif op == "-":
        print(a - b)

    elif op == "*":
        print(a * b)

    elif op == "/":
        print(a / b)

    else:
        print("Invalid Operator")

except ValueError:
    print("Invalid Number")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
try:
    file = open("data.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File Not Found")

except PermissionError:
    print("Permission Denied")
```

```python
class InsufficientBalanceError(Exception):
    pass


balance = 5000

try:
    amount = int(input("Withdraw Amount: "))

    if amount > balance:
        raise InsufficientBalanceError("Insufficient Balance")

    balance -= amount

    print("Remaining Balance:", balance)

except InsufficientBalanceError as e:
    print(e)
```

```python
numbers = input("Enter numbers separated by space: ").split()

result = []

for num in numbers:
    try:
        result.append(int(num))

    except ValueError:
        print(num, "is not a valid integer.")

print(result)
```

```python
try:
    name = input("Enter Name: ")

    age = int(input("Enter Age: "))

    marks = float(input("Enter Marks: "))

    if age < 0:
        raise ValueError("Age cannot be negative.")

    print("\nStudent Details")
    print("Name :", name)
    print("Age :", age)
    print("Marks :", marks)

except ValueError as e:
    print("Error:", e)

except Exception as e:
    print("Unexpected Error:", e)

finally:
    print("Program Ended")
```
