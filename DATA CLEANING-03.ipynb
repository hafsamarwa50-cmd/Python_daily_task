```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/user/Downloads/Customer_datasets - Customer_datasets.csv")

df["Age"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = (
    (df["Age"] < lower_limit) |
    (df["Age"] > upper_limit)
)

outliers_removed = outliers.sum()

df = df[~outliers]

print("Number of outliers removed:", outliers_removed)
print("Rows remaining:", len(df))
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/user/Downloads/Customer_datasets - Customer_datasets.csv")

df["Salary"] = (
    df["Salary"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

mean_salary = df["Salary"].mean()
median_salary = df["Salary"].median()

print("Mean Salary:", mean_salary)
print("Median Salary:", median_salary)

print("\nThe mean is generally more affected by outliers than the median.")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/user/Downloads/Customer_datasets - Customer_datasets.csv")

original_gender = df["Gender"].copy()

df["Gender"] = (
    df["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["Gender"] = df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "o": "Other",
    "other": "Other"
})

df["Gender"] = df["Gender"].fillna("Other")

changed_records = (
    original_gender.astype(str).str.strip().str.lower()
    !=
    df["Gender"].astype(str).str.lower()
).sum()

print("Records changed:", changed_records)

print("\nFinal Gender values:")
print(df["Gender"].value_counts())
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/user/Downloads/Customer_datasets - Customer_datasets.csv")

email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

df["Invalid_Email"] = ~df["Email"].astype(str).str.match(
    email_pattern,
    na=False
)

invalid_emails = df["Invalid_Email"].sum()

print("Number of invalid emails:", invalid_emails)

print("\nInvalid emails:")
print(df.loc[df["Invalid_Email"], "Email"])
```
