```python
import pandas as pd

df = pd.read_csv("/Users/user/Downloads/customer_analytics - customer_analytics.csv")

df.head(10)
df.tail(10)
df.sample(10)
df.info()

print(df.isnull().sum())
print(df.duplicated().sum())

df = df.drop_duplicates()

# Numeric cleaning
for col in ["Age", "Annual_Income", "Account_Balance",
            "Credit_Score", "Monthly_Spend",
            "Transactions", "Satisfaction_Score"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Annual_Income"] = df["Annual_Income"].fillna(df["Annual_Income"].median())
df["Account_Balance"] = df["Account_Balance"].fillna(df["Account_Balance"].mean())
df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].median())
df["Monthly_Spend"] = df["Monthly_Spend"].fillna(df["Monthly_Spend"].median())
df["Transactions"] = df["Transactions"].fillna(df["Transactions"].median())
df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(df["Satisfaction_Score"].mean())

# Date
df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
df["Join_Date"] = df["Join_Date"].fillna(df["Join_Date"].median())

# Categories
df["City"] = df["City"].str.title()
df["Segment"] = df["Segment"].fillna(df["Segment"].mode()[0])

df["Active"] = df["Active"].replace({
    "Y": "Yes", "YES": "Yes", "yes": "Yes", "1": "Yes", 1: "Yes",
    "N": "No", "NO": "No", "no": "No", "0": "No", 0: "No"
})

# Rename
df = df.rename(columns={
    "Annual_Income": "AnnualIncome",
    "Account_Balance": "AccountBalance",
    "Credit_Score": "CreditScore",
    "Tenure_Years": "Tenure",
    "Join_Date": "CustomerJoinDate",
    "Risk_Level": "RiskLevel",
    "Monthly_Spend": "MonthlySpend",
    "Satisfaction_Score": "SatisfactionScore"
})

# Drop ID
df = df.drop("Customer_ID", axis=1)

df.info()
df.describe()
```

```python
# Analytical features
df["Income_Lakhs"] = df["AnnualIncome"] / 100000
df["MonthlyIncome"] = df["AnnualIncome"] / 12
df["Income_After_Spending"] = df["MonthlyIncome"] - df["MonthlySpend"]
df["Balance_Income_Ratio"] = df["AccountBalance"] / df["AnnualIncome"].replace(0, pd.NA)

# Final validation
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:")
print(df.isnull().sum())

print("\nFinal Data Types:")
df.info()

print("\nDescriptive Statistics:")
print(df.describe())

print("\nCity Distribution:")
print(df["City"].value_counts())

print("\nSegment Distribution:")
print(df["Segment"].value_counts())

print("\nActive Distribution:")
print(df["Active"].value_counts())

print("\nRisk Level Distribution:")
print(df["RiskLevel"].value_counts())

print("\nFirst 10 Rows:")
display(df.head(10))

print("\nLast 10 Rows:")
display(df.tail(10))

print("\nRandom 10 Rows:")
display(df.sample(10))
```
