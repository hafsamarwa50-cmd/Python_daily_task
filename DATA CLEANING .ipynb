# Q25 — Convert Total Stops to Integer

```python
import pandas as pd

df = pd.read_csv(
    "/Users/user/Downloads/Data_Train.xlsx - Sheet1.csv"
)

df["Total_Stops"] = df["Total_Stops"].replace(
    "non-stop",
    "0"
)

df["Total_Stops"] = pd.to_numeric(
    df["Total_Stops"]
    .astype("string")
    .str.extract(r"(\d+)")[0],
    errors="coerce"
)

df["Total_Stops"] = (
    df["Total_Stops"]
    .fillna(0)
    .astype(int)
)

print(df["Total_Stops"].value_counts())
print(df["Total_Stops"].dtype)
```

# Q26 — Split Route

```python
import pandas as pd

df = pd.read_csv(
    "/Users/user/Downloads/Data_Train.xlsx - Sheet1.csv"
)

route_columns = [
    "Route_1",
    "Route_2",
    "Route_3",
    "Route_4",
    "Route_5"
]

route_split = df["Route"].astype("string").str.split(
    r"\s*(?:→|->|>)\s*",
    expand=True
)

# Keep only 5 route columns
route_split = route_split.iloc[:, :5]

# Add missing columns if needed
while route_split.shape[1] < 5:
    route_split[route_split.shape[1]] = None

route_split.columns = route_columns

df[route_columns] = route_split

print(df[route_columns].head())
```
