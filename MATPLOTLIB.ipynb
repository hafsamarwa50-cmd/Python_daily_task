```python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch", "Headphones"]
revenue = [50000, 80000, 40000, 30000, 60000]

bars = plt.bar(products, revenue)

plt.title("Product Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        str(value),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

products = ["A", "B", "C", "D", "E"]
sales = [50, 70, 40, 80, 60]

age = [22, 25, 28, 30, 35]
salary = [30000, 35000, 40000, 45000, 55000]

values = np.random.randn(100)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Line Chart")

plt.subplot(2, 2, 2)
plt.bar(products, sales)
plt.title("Bar Chart")

plt.subplot(2, 2, 3)
plt.scatter(age, salary)
plt.title("Scatter Plot")

plt.subplot(2, 2, 4)
plt.hist(values, bins=10)
plt.title("Histogram")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [20000, 25000, 22000, 30000, 28000, 35000]

plt.figure(figsize=(10, 5))

plt.plot(months, sales, marker="o")

max_sales = max(sales)
max_index = sales.index(max_sales)
max_month = months[max_index]

plt.annotate(
    "Maximum Sales",
    xy=(max_month, max_sales),
    xytext=(max_month, max_sales + 3000),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.show()

print("Maximum Sales Month:", max_month)
print("Maximum Sales:", max_sales)
```

```python
import matplotlib.pyplot as plt

age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45]
income = [50000, 75000, 90000, 110000, 60000,
          125000, 80000, 150000, 95000, 130000]

normal_age = []
normal_income = []

high_age = []
high_income = []

for i in range(len(income)):
    if income[i] > 100000:
        high_age.append(age[i])
        high_income.append(income[i])
    else:
        normal_age.append(age[i])
        normal_income.append(income[i])

plt.scatter(normal_age, normal_income, label="Income ≤ ₹1,00,000")
plt.scatter(high_age, high_income, marker="*", s=150,
            label="Income > ₹1,00,000")

plt.title("Age vs Income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

online = [20, 25, 30, 35, 40]
offline = [15, 20, 18, 25, 30]
corporate = [10, 15, 20, 18, 25]

plt.bar(months, online, label="Online")
plt.bar(months, offline, bottom=online, label="Offline")

bottom_values = [online[i] + offline[i] for i in range(len(months))]

plt.bar(months, corporate, bottom=bottom_values, label="Corporate")

plt.title("Monthly Sales by Channel")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

actual = [80, 90, 75, 110, 95, 120, 100, 130, 115, 140, 125, 150]
target = [85, 85, 85, 100, 100, 110, 110, 120, 120, 130, 130, 140]

plt.figure(figsize=(10, 5))

plt.plot(months, actual, marker="o", label="Actual Sales")
plt.plot(months, target, linestyle="--", label="Target Sales")

for i in range(len(months)):
    if actual[i] < target[i]:
        plt.scatter(months[i], actual[i], marker="x", s=100)

plt.title("Actual Sales vs Target Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [20000, 25000, 22000, 30000, 28000, 35000]

categories = ["Electronics", "Clothing", "Food", "Furniture"]
category_sales = [50000, 40000, 30000, 20000]

age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 50, 55]

discount = [5, 10, 15, 20, 25, 30, 10, 15, 20, 25]
revenue_discount = [50000, 48000, 45000, 42000, 40000,
                    38000, 47000, 44000, 41000, 39000]

plt.figure(figsize=(12, 10))

# 1. Monthly revenue
plt.subplot(2, 2, 1)
plt.plot(months, revenue, marker="o", label="Revenue")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.legend()

# 2. Category sales
plt.subplot(2, 2, 2)
plt.bar(categories, category_sales)
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=30)

# 3. Customer age histogram
plt.subplot(2, 2, 3)
plt.hist(age, bins=6)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.grid(True)

# 4. Revenue vs discount
plt.subplot(2, 2, 4)
plt.scatter(discount, revenue_discount)
plt.title("Revenue vs Discount")
plt.xlabel("Discount (%)")
plt.ylabel("Revenue")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# plt.plot()       → Line chart
# plt.bar()        → Vertical bar chart
# plt.barh()       → Horizontal bar chart
# plt.scatter()    → Scatter plot
# plt.hist()       → Histogram
# plt.pie()        → Pie chart
# plt.subplot()    → Multiple plots
# plt.title()      → Title
# plt.xlabel()     → X-axis label
# plt.ylabel()     → Y-axis label
# plt.legend()     → Legend
# plt.grid()       → Grid
# plt.show()       → Display graph
```
