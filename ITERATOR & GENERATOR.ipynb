```python
class Numbers:

    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Numbers()

for i in obj:
    print(i)
```

```python
class Reverse:

    def __iter__(self):
        self.num = 10
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration

obj = Reverse()

for i in obj:
    print(i)
```

```python
class Word:

    def __iter__(self):
        self.word = "GENERATOR"
        self.index = 0
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration

obj = Word()

for i in obj:
    print(i)
```

```python
class Table:

    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num * 5
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Table()

for i in obj:
    print(i)
```

```python
def numbers():
    for i in range(1, 11):
        yield i

gen = numbers()

for i in gen:
    print(i)
```

# ADVANCED ITERATOR PROBLEMS:

```python
class NaturalNumbers:

    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 15:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = NaturalNumbers()

for i in obj:
    print(i)
```

```python
class MultiplesOfThree:

    def __iter__(self):
        self.num = 3
        return self

    def __next__(self):
        if self.num <= 30:
            value = self.num
            self.num += 3
            return value
        else:
            raise StopIteration

obj = MultiplesOfThree()

for i in obj:
    print(i)
```

```python
class Alphabet:

    def __iter__(self):
        self.ch = ord('A')
        return self

    def __next__(self):
        if self.ch <= ord('Z'):
            letter = chr(self.ch)
            self.ch += 1
            return letter
        else:
            raise StopIteration

obj = Alphabet()

for i in obj:
    print(i)
```

```python
class Vowels:

    def __iter__(self):
        self.text = "Artificial Intelligence"
        self.index = 0
        return self

    def __next__(self):
        while self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1

            if ch.lower() in "aeiou":
                return ch

        raise StopIteration

obj = Vowels()

for i in obj:
    print(i)
```

```python
class Fibonacci:

    def __iter__(self):
        self.a = 0
        self.b = 1
        self.count = 0
        return self

    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration

obj = Fibonacci()

for i in obj:
    print(i)
```

```python
def fibonacci():
    a = 0
    b = 1

    for i in range(15):
        yield a
        a, b = b, a + b

gen = fibonacci()

for i in gen:
    print(i)
```

```python
def prime_numbers():

    for num in range(2, 101):

        prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                prime = False
                break

        if prime:
            yield num

gen = prime_numbers()

for i in gen:
    print(i)
```

```python
num = int(input("Enter Number: "))

def factors():

    for i in range(1, num + 1):
        if num % i == 0:
            yield i

gen = factors()

print("Factors are:")

for i in gen:
    print(i)
```

```python
def countdown():

    for i in range(20, 0, -1):
        yield i

gen = countdown()

for i in gen:
    print(i)
```

```python
def multiples_of_five():

    num = 5

    while True:
        yield num
        num += 5

gen = multiples_of_five()

for i in range(15):
    print(next(gen))
```
