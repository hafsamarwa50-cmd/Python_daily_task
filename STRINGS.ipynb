```python
text = input("Enter a string: ")

print("Last 4 characters:", text[-4:])
```

```python
name = input("Enter full name: ")

space = name.find(" ")

first_name = name[:space]

last_name = name[space + 1:]

print("First Name:", first_name)
print("Last Name:", last_name)
```

# CASE CONVERSION QUESTIONS:

```python
text = input("Enter a string: ")

print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title Case:", text.title())
print("Capitalized:", text.capitalize())
```

```python
sentence = input("Enter a sentence: ")

print("Original:", sentence)
print("Title Case:", sentence.title())
```

```python
text = input("Enter a mixed case string: ")

print("Original:", text)
print("Lower:", text.lower())
print("Upper:", text.upper())
print("After methods:", text)
```

# STRING CHECK METHODS:

```python
text = input("Enter a string: ")

if text.isalpha():
    print("Contains only alphabets")
else:
    print("Contains other characters")
```

```python
text = input("Enter a value: ")

if text.isdigit():
    print("Only digits")
else:
    print("Not only digits")
```

```python
text = input("Enter a string: ")

if text.isalnum():
    print("Alphanumeric")
else:
    print("Not Alphanumeric")
```

```python
text = input("Enter a string: ")

if text.islower():
    print("String is lowercase")
else:
    print("String is not lowercase")
```

```python
password = input("Enter password: ")

if password.isalnum() and len(password) >= 8:
    print("Valid Password")
else:
    print("Invalid Password")
```

# FIND() ,INDEX():

```python
text = input("Enter a string: ")

ch = input("Enter a character: ")

position = text.find(ch)

print("Position:", position)
```

```python
text = input("Enter a string: ")

word = input("Enter a word: ")

position = text.find(word)

print(position)
```

```python
sentence = input("Enter a sentence containing Python: ")

position = sentence.index("Python")

print("Position:", position)
```

```python
text = input("Enter a string: ")

ch = input("Enter a character that is not present: ")

print("Using find():", text.find(ch))

print("Using index():")
print(text.index(ch))
```

# MIXED:

```python
name = input("Enter your full name: ")

upper_name = name.upper()

space = name.find(" ")

first_name = name[:space]

last_name = name[space + 1:]

print("Uppercase:", upper_name)
print("Position of Space:", space)
print("First Name:", first_name)
print("Last Name:", last_name)
```

```python
mobile = input("Enter mobile number: ")

if mobile.isdigit() and len(mobile) == 10:
    print("Valid Mobile Number")
else:
    print("Invalid Mobile Number")
```

```python
text = input("Enter a string: ")

reverse = text[::-1]

print("Reversed String:", reverse)

print("Title Case:", text.title())

if text == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```
