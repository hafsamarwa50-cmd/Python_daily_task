# Challenge 20 – Final Interview Project

```python
import pandas as pd

df = pd.read_csv("/Users/user/Downloads/Task_dataset - Task_dataset.csv")

# Check information
df.info()

# Find missing values
print(df.isnull().sum())

# Remove duplicates
df = df.drop_duplicates()

# Rename columns
df = df.rename(columns={
    "CustomerID": "Customer_ID",
    "AnnualIncome": "Annual_Income",
    "CreditScore": "Credit_Score",
    "Target": "Customer_Status"
})

# Convert Age
df["Age"] = df["Age"].fillna(
    df["Age"].median()
).astype(int)

# Convert AnnualIncome
df["Annual_Income"] = df["Annual_Income"].fillna(
    df["Annual_Income"].median()
).astype(float)

# Convert Balance
df["Balance"] = df["Balance"].fillna(
    df["Balance"].mean()
).astype(float)

# Remove Segment and CustomerID
df = df.drop(columns=["Segment", "Customer_ID"])

# Create new columns
df["TotalMoney"] = df["Balance"] + df["Annual_Income"]

df["AgeAfter5Years"] = df["Age"] + 5

df["MonthlyIncome"] = df["Annual_Income"] / 12

df["CreditScoreDouble"] = df["Credit_Score"] * 2

# Head
print("HEAD")
print(df.head())

# Tail
print("\nTAIL")
print(df.tail())

# Sample
print("\nSAMPLE")
print(df.sample(10))

# Describe
print("\nDESCRIBE")
print(df.describe())

# Target counts
print("\nTARGET COUNTS")
print(df["Customer_Status"].value_counts())

# Final information
print("\nFINAL INFORMATION")
df.info()
```
