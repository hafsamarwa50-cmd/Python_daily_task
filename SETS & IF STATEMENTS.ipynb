# SETS QUESTIONS:

```python
s = {1, 2, 3, 2, 4, 1}

print(s)
```

```python
s = set()

s.add(10)
s.add(20)
s.add(30)

print(s)
```

```python
s = {10, 20, 30}

s.update([40, 50, 60])

print(s)
```

```python
s = {10, 20, 30, 40}

s.remove(20)
print(s)

s.discard(30)
print(s)
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union =", A | B)
print("Intersection =", A & B)
print("Difference =", A - B)
```

```python
s = {10, 20, 30, 40}

if 25 in s:
    print("Present")
else:
    print("Not Present")
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

s = set(numbers)

print(s)
print("Length =", len(s))
```

```python
s = set()

for i in range(5):
    num = int(input("Enter number: "))
    s.add(num)

print("Unique Numbers:", s)
```

# EXTRA HARD QUESTIONS:

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a == b == c:
    print("All three numbers are equal")
else:
    if a >= b and a >= c:
        print("Largest =", a)
    elif b >= a and b >= c:
        print("Largest =", b)
    else:
        print("Largest =", c)

    print("All three numbers are not equal")
```

```python
year = int(input("Enter year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:

    if a == b == c:
        print("Equilateral Triangle")

    elif a == b or b == c or a == c:
        print("Isosceles Triangle")

    else:
        print("Scalene Triangle")

else:
    print("Invalid Triangle")
```

```python
s1 = int(input("Enter Subject 1 Marks: "))
s2 = int(input("Enter Subject 2 Marks: "))
s3 = int(input("Enter Subject 3 Marks: "))
s4 = int(input("Enter Subject 4 Marks: "))
s5 = int(input("Enter Subject 5 Marks: "))

percentage = (s1 + s2 + s3 + s4 + s5) / 5

print("Percentage =", percentage)

if percentage >= 90:
    print("Grade: A+")

elif percentage >= 80:
    print("Grade: A")

elif percentage >= 70:
    print("Grade: B")

elif percentage >= 60:
    print("Grade: C")

else:
    print("Fail")

if percentage >= 75:
    print("Passed with Distinction")
```
