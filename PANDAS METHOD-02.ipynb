# MINI PROJECT:

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/Users/user/Downloads/Bank_customer_churn.csv")

# Dataset information
print(df.info())

# Missing values
print(df.isnull().sum())

# Remove duplicates
df = df.drop_duplicates()

# Rename columns
df = df.rename(columns={
    "CustomerId": "Customer_ID",
    "CreditScore": "Credit_Score",
    "EstimatedSalary": "Salary",
    "Exited": "Target"
})

# Convert datatypes
df["Age"] = df["Age"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Target"] = df["Target"].astype(str)

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

# Remove columns
df = df.drop(columns=["RowNumber", "Surname"])

# First 10 rows
print(df.head(10))

# Statistical summary
print(df.describe())

# Target counts
print(df["Target"].value_counts())

# Final information
print(df.info())
```
