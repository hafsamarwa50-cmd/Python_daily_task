```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number =", reverse)
```

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not a Prime Number")
else:
    i = 2
    prime = True

    while i * i <= num:
        if num % i == 0:
            prime = False
            break
        i += 1

    if prime:
        print("Prime Number")
    else:
        print("Not a Prime Number")
```

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1
count = 1

while count <= n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```

```python
correct_password = "python123"

while True:
    password = input("Enter Password: ")

    if password == correct_password:
        print("Access Granted")
        break
    else:
        print("Wrong Password. Try Again.")
```

```python
num = int(input("Enter a number: "))

total = 0

while num > 0:
    digit = num % 10
    total += digit
    num = num // 10

print("Sum of digits =", total)
```

# NESTED FOR LOOP:

```python
for i in range(1, 6):
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```

```python
for i in range(1, 6):
    print("Table of", i)

    for j in range(1, 11):
        print(i, "x", j, "=", i * j)

    print()
```

```python
for num in range(2, 101):

    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num)
```

```python
n = int(input("Enter size of square: "))

for i in range(1, n + 1):
    for j in range(1, n + 1):

        if i == 1 or i == n or j == 1 or j == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

# BREAK AND CONTINUE:

```python
for i in range(1, 21):
    if i == 13:
        break
    print(i)
```

```python
for i in range(1, 31):
    if i % 5 == 0:
        continue
    print(i)
```

```python
for i in range(5):
    num = int(input("Enter a number: "))

    if num < 0:
        print("Stopped due to negative number.")
        break

    print("You entered:", num)
```

```python
for i in range(1, 21):
    if i % 2 != 0:
        continue

    print(i)
```

```python
for i in range(1, 101):
    if i % 3 == 0 and i % 7 == 0:
        print(i)
        break
```

```python
total = 0

while True:
    num = int(input("Enter a number: "))
    total += num

    if total > 100:
        break

print("Final Sum =", total)
```

```python
for i in range(1, 21):

    if i == 5 or i == 10 or i == 15:
        continue

    print(i)
```

```python
for i in range(1, 4):

    for j in range(1, 6):

        if j == 3:
            break

        print("i =", i, "j =", j)

    print()
```

```python
while True:
    word = input("Enter a word: ")

    if word.lower() == "exit":
        print("Loop Ended")
        break

    print("You entered:", word)
```

```python
for num in range(10, 51):

    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if not prime:
        continue

    print(num)
```
