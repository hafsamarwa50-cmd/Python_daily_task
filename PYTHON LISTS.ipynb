```python
students = ["Rahul", "Aman", "Priya", "Neha", "Kiran"]

print(students)
```

```python
numbers = [10, 20, 30, 40, 50]

print(numbers)
print(type(numbers))
print(len(numbers))
```

# MEDIUM LEVEL:

```python
lst = [10, 15, 25, 30, 40]

print(lst.index(25))
```

```python
lst = [7, 2, 7, 5, 7, 8]

print(lst.count(7))
```

```python
lst = [45, 12, 78, 23, 56]

lst.sort()

print(lst)
```

```python
lst = [45, 12, 78, 23, 56]

lst.sort(reverse=True)

print(lst)
```

```python
fruits = ["apple", "banana", "mango", "orange"]

fruits.reverse()

print(fruits)
```

```python
original = [1, 2, 3, 4]

copy_list = original.copy()

print("Original:", original)
print("Copy:", copy_list)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

print(words)
```

# HARD / MIXED LEVEL:

```python
numbers = [10, 40, 20, 60, 30]

numbers.append(70)
numbers.append(15)
numbers.append(50)

numbers.sort(reverse=True)

print(numbers)
```

```python
lst = ["Java", "C", "C++"]

lst.insert(0, "Python")

lst.pop()

print(lst)
print("Length:", len(lst))
```

```python
cities = []

for i in range(5):
    city = input("Enter city name: ")
    cities.append(city)

cities.reverse()

print("Cities in reverse order:")
print(cities)
```

```python
data = [10, 20, 30, 20, 40, 20]

while 20 in data:
    data.remove(20)

print(data)
```

```python
marks = [78, 95, 62, 88, 70]

marks.sort()

print("Sorted Marks:", marks)
print("Lowest Mark:", marks[0])
print("Highest Mark:", marks[-1])
```
