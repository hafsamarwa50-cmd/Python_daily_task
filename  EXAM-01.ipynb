```python
### ============================================
# BANKING SYSTEM (PART 1)
# Custom Exceptions + Abstract Class +
# Savings Account
# ============================================

from abc import ABC, abstractmethod


# ============================================
# Custom Exceptions
# ============================================

class InsufficientBalanceError(Exception):
    pass


class InvalidAmountError(Exception):
    pass


class AccountNotFoundError(Exception):
    pass


class DailyLimitExceededError(Exception):
    pass


# ============================================
# Abstract Class
# ============================================

class Account(ABC):

    total_accounts = 0
    next_account_no = 1001

    def __init__(self, name, balance):

        self._account_no = Account.next_account_no
        Account.next_account_no += 1

        self._name = name
        self._balance = balance

        Account.total_accounts += 1

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def get_balance(self):
        pass

    def show_account_info(self):
        print("\n========== ACCOUNT DETAILS ==========")
        print("Account Number :", self._account_no)
        print("Account Holder :", self._name)
        print("Balance        :", self.get_balance())

    @classmethod
    def show_total_accounts(cls):
        print("\nTotal Accounts :", cls.total_accounts)


# ============================================
# Savings Account
# ============================================

class SavingsAccount(Account):

    INTEREST_RATE = 0.04
    DAILY_LIMIT = 25000
    MIN_BALANCE = 1000

    def __init__(self, name, balance):

        super().__init__(name, balance)

        self.__daily_withdrawn = 0

    # Getter
    def get_balance(self):
        return self._balance

    # Setter
    def set_balance(self, amount):

        if amount < 0:
            raise InvalidAmountError("Balance cannot be negative.")

        self._balance = amount

    # Deposit
    def deposit(self, amount):

        if amount <= 0:
            raise InvalidAmountError(
                "Deposit amount must be greater than zero."
            )

        self.set_balance(self.get_balance() + amount)

        print("Deposit Successful.")
        print("Current Balance :", self.get_balance())

    # Withdraw
    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError(
                "Withdrawal amount must be greater than zero."
            )

        if self.__daily_withdrawn + amount > self.DAILY_LIMIT:
            raise DailyLimitExceededError(
                "Daily withdrawal limit is ₹25000."
            )

        if self.get_balance() - amount < self.MIN_BALANCE:
            raise InsufficientBalanceError(
                "Minimum balance of ₹1000 must be maintained."
            )

        self.set_balance(self.get_balance() - amount)

        self.__daily_withdrawn += amount

        print("Withdrawal Successful.")
        print("Remaining Balance :", self.get_balance())

    # Interest
    def add_interest(self):

        interest = self.get_balance() * self.INTEREST_RATE

        self.set_balance(self.get_balance() + interest)

        print("Interest Added :", interest)
        print("Current Balance :", self.get_balance())
        # ============================================
# Current Account
# ============================================

class CurrentAccount(Account):

    OVERDRAFT_LIMIT = 10000

    def __init__(self, name, balance):
        super().__init__(name, balance)

    # Getter
    def get_balance(self):
        return self._balance

    # Setter
    def set_balance(self, amount):
        self._balance = amount

    # Deposit
    def deposit(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than 0.")

        self.set_balance(self.get_balance() + amount)

        print("Deposit Successful.")
        print("Current Balance :", self.get_balance())

    # Withdraw
    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than 0.")

        # Allow overdraft up to ₹10,000
        if self.get_balance() - amount < -self.OVERDRAFT_LIMIT:
            raise InsufficientBalanceError(
                "Overdraft limit of ₹10000 exceeded."
            )

        self.set_balance(self.get_balance() - amount)

        print("Withdrawal Successful.")
        print("Remaining Balance :", self.get_balance())
        # ============================================
# Bank Class
# ============================================

class Bank:

    def __init__(self):
        self.accounts = {}

    # ----------------------------
    # Create Account
    # ----------------------------
    def create_account(self, acc_type, name, balance):

        if balance <= 0:
            raise InvalidAmountError(
                "Opening balance must be greater than 0."
            )

        if acc_type == 1:
            account = SavingsAccount(name, balance)

        elif acc_type == 2:
            account = CurrentAccount(name, balance)

        else:
            print("Invalid Account Type")
            return

        self.accounts[account._account_no] = account

        print("\nAccount Created Successfully!")
        print("Account Number :", account._account_no)

    # ----------------------------
    # Find Account
    # ----------------------------
    def find_account(self, account_no):

        if account_no not in self.accounts:
            raise AccountNotFoundError("Account does not exist.")

        return self.accounts[account_no]

    # ----------------------------
    # Deposit
    # ----------------------------
    def deposit(self, account_no, amount):

        account = self.find_account(account_no)

        account.deposit(amount)

    # ----------------------------
    # Withdraw
    # ----------------------------
    def withdraw(self, account_no, amount):

        account = self.find_account(account_no)

        account.withdraw(amount)

    # ----------------------------
    # Check Balance
    # ----------------------------
    def check_balance(self, account_no):

        account = self.find_account(account_no)

        print("Current Balance :", account.get_balance())

    # ----------------------------
    # Show Account Information
    # ----------------------------
    def show_info(self, account_no):

        account = self.find_account(account_no)

        account.show_account_info()
        # ============================================
# Main Program
# ============================================

bank = Bank()

while True:

    print("\n========== Welcome to XYZ Bank ==========")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Show Account Info")
    print("6. Show Total Accounts")
    print("7. Exit")

    try:

        choice = int(input("Enter Choice: "))

        # -----------------------------------
        # Create Account
        # -----------------------------------
        if choice == 1:

            print("\n1. Savings Account")
            print("2. Current Account")

            acc_type = int(input("Enter Account Type: "))
            name = input("Enter Name: ")
            balance = float(input("Enter Opening Balance: "))

            bank.create_account(acc_type, name, balance)

        # -----------------------------------
        # Deposit
        # -----------------------------------
        elif choice == 2:

            account_no = int(input("Enter Account Number: "))
            amount = float(input("Enter Deposit Amount: "))

            bank.deposit(account_no, amount)

        # -----------------------------------
        # Withdraw
        # -----------------------------------
        elif choice == 3:

            account_no = int(input("Enter Account Number: "))
            amount = float(input("Enter Withdraw Amount: "))

            bank.withdraw(account_no, amount)

        # -----------------------------------
        # Check Balance
        # -----------------------------------
        elif choice == 4:

            account_no = int(input("Enter Account Number: "))

            bank.check_balance(account_no)

        # -----------------------------------
        # Show Account Information
        # -----------------------------------
        elif choice == 5:

            account_no = int(input("Enter Account Number: "))

            bank.show_info(account_no)

        # -----------------------------------
        # Total Accounts
        # -----------------------------------
        elif choice == 6:

            Account.show_total_accounts()

        # -----------------------------------
        # Exit
        # -----------------------------------
        elif choice == 7:

            print("\nThank You for Banking with XYZ Bank.")
            break

        else:
            print("Invalid Choice.")

    # ---------------------------------------
    # Exception Handling
    # ---------------------------------------

    except InvalidAmountError as e:
        print("Error:", e)

    except InsufficientBalanceError as e:
        print("Error:", e)

    except DailyLimitExceededError as e:
        print("Error:", e)

    except AccountNotFoundError as e:
        print("Error:", e)

    except ValueError:
        print("Please enter valid numeric input.")

    except Exception as e:
        print("Unexpected Error:", e)

    else:
        print("Transaction Completed Successfully.")

    finally:
        print("Thank You.")
```
